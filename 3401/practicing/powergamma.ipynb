```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
def image_show(img) :
    cv2.imshow("xxxx",img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    cv2.waitKey(1)
```

```python
def calculate_hist(img,foldername=None,name=None) :
    hist = cv2.calcHist([img],[0],None,[256],[0,256]) 
    plt.figure()
    plt.plot(hist)
    if name and foldername :
        plt.savefig(foldername+"/"+name)
    plt.show()
    return hist
```

```python
img = cv2.imread('../image/A1.tif',0)
image_show(img)
hist = calculate_hist(img)
```

```python
img = img.astype('float')/np.max(img)
img = (img**0.5*255).astype('uint8')
image_show(img)
hist = calculate_hist(img)
```
