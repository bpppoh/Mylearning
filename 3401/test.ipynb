```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros((256,256)).astype('uint8')
```

```python
for i in range(256):
    for j in range(256) :
        img[i,j] = 255-j
```

```python
plt.imshow(img,cmap='gray')
plt.axis('off')
plt.show()
```

```python
hist = cv2.calcHist([img],[0],None,[256],[0,256])
plt.plot(hist)
plt.show()
```
