```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('../image/bank4.jpg' , 0)
print(f"type(img) : {type(img)}")
```

```python
hist = cv2.calcHist([img],[0],None,[256],[0,256])
print(f"hist.shape : {hist.shape}")
```

```python
plt.plot(hist)
plt.xlabel("Intensity")
plt.ylabel("Pixel amount")
plt.show()
```

```python
thresh = 128
out = np.zeros_like(img,dtype='uint8')
x = img > thresh
out[x] = 255
```

```python
def imageShow(image) :
    cv2.imshow('xxxx',image)
    cv2.waitKey()
    cv2.destroyAllWindows()
    for i in range(0,5) :
        cv2.waitKey(1)
```

```python
imageShow(img)
imageShow(out)
```

```python
ret, bin_img = cv2.threshold(img, thresh, 255, cv2.THRESH_BINARY)
print(ret)
imageShow(bin_img)
```
