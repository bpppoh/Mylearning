```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('../image/cameraman.tif' , 0)
print(type(img))
print("cv2.IMREAD_GRAYSCALE : ",cv2.IMREAD_GRAYSCALE)
```

```python
# create list of 0 , 256 elements
inten = [0] * 256
print(inten)
```

```python
# create array from list that recently created
inten = np.array(inten , dtype = 'int')
print(inten)
print(type(inten))
```

```python
m,n = img.shape # m got row , n got column of img
inten = np.array([0]*256 , dtype="int") # shortcut of making array
for i in range(m) :
    for j in range(n) :
        inten[img[i,j]] += 1

# แต่ละ index ใน inten จะเก็บจำนวนจุดภาพในแตแ่ละความเข้ม
# print(inten)
print(f"m : {m}")
print(f"n : {n}")
print(f"m*n : {m*n}")
print(f"inten.sum() : {inten.sum()}")
print(f"inten.shape : {inten.shape}")
```

```python
plt.plot(inten)
plt.title("intensity")
plt.xlabel("Intensity")
plt.ylabel("Frequency")
plt.show()
```
