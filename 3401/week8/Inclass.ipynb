```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread('../image/cameraman.tif',0)
plt.imshow(img , cmap='gray')
plt.axis('off')
plt.show()
```

```python
def centroid_image(img) :
    r , c = img.shape
    return (r//2,c//2)
```

```python
cen = centroid_image(img)
print(cen)
```

```python
def img_translation1(img,tx,ty) :
    out = np.zeros_like(img,dtype='uint8')
    ts = np.identity(3,dtype=float)
    ts[2,0] = tx
    ts[2,1] = ty
    row , col = img.shape
    for x in range(col) :
        for y in range(row) :
            xy = np.array([x,y,1],dtype=float)
            new_xy = np.matmul(xy,ts)
            new_x = int(new_xy[0])
            new_y = int(new_xy[1])
            if 0 <= new_x < col and 0 <= new_y < row :
                out[new_y,new_x] = img[y,x]
    return out.astype('uint8')
```

```python
newImg = img_translation1(img,50,50) 
plt.imshow(newImg , cmap='gray')
plt.axis('off')
plt.show()
```
