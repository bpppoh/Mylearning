```python
def RGB2CMYK(img) :
    img = img/255.
    r = img[:,:,2]
    g = img[:,:,1]
    b = img[:,:,0]
    
    c = 1 - r
    m = 1 - g
    y = 1 - b
    k = np.minimum(np.minimum(c,m),y)
    
    denominator = 1 - k
    c = (1-c)/denominator
    m = (1-m)/denominator
    y = (1-y)/denominator
    
    return np.dstack((c,m,y,k)*255).astype('uint8')
```

```python
cmykImg = RGB2CMYK(img)
print(cmykImg.shape)
```
