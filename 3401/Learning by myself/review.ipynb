```python
def intermean(hist , thr) :
    p = hist/hist.sum()
    
    indices = np.arange(256)
    i0 = indices[:thr]
    i1 = indices[thr:]
    
    p0 = p[:thr]
    p1 = p[thr:]
    
    w0 = p0.sum() + 1e-8
    w1 = p1.sum() + 1e-8
    
    u0 = (i0*p0).sum() / w0
    u1 = (i1*p1).sum() / w1
    
    if u0 == 0 :
        thr = u0
    elif u1 == 0 :
        thr = u1
    else :
        thr = (u0 + u1) / 2
    
    return int(np.round(thr))
```
