```python
import numpy as np
```

```python
img = np.zeros_like([256,256],dtype='float')
```

```python
def saturation(np_array) :
    norm = np_array/255.
    c_max = np.amax(norm)
    c_min = np.amin(norm)
    delta = c_max - c_min
    saturation = delta / c_max
    return saturation
```

```python
img1 = np.array((255,55,255))
img2 = np.array((0,0,255))
```

```python
s1 = saturation(img1)
s2 = saturation(img2)
print(s1)
print(s2)
```
