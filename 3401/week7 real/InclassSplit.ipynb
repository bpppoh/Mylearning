```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt
```

```python
def pltshow(img) :
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```

```python
img = cv2.imread('../image/gray3.png')
pltshow(img)
```

```python
def split4(img) :
    h_half = img.shape[0] // 2
    w_half = img.shape[1] // 2
    
    img_11 = img[:h_half,:w_half]
    img_12 = img[:h_half,w_half:]
    img_21 = img[h_half:,:w_half]
    img_22 = img[h_half:,w_half:]
    
    return [img_11,img_12,img_21,img_22]
```

```python
x = split4(img)
pltshow(x[0])
```
