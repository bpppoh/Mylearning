```python
import os
import shutil
```

```python
path = "./organized_dataset"
dataPath = "./messy_data"
imagesFolderPath = os.path.join(path,"images")
labelsFolderPath = os.path.join(path,"labels")

if os.path.exists(imagesFolderPath) :
    shutil.rmtree(imagesFolderPath)
if os.path.exists(labelsFolderPath) :
    shutil.rmtree(labelsFolderPath)
os.makedirs(imagesFolderPath,exist_ok=True)
os.makedirs(labelsFolderPath,exist_ok=True)

print(f"Is images folder created ? : {os.path.exists(imagesFolderPath)}")
print(f"Is labels folder created ? : {os.path.exists(labelsFolderPath)}")
```

```python
os.path.exists(dataPath)

datas = os.listdir(dataPath)
print(datas)

for data in datas :
    if data.endswith(('.jpg','.png')) :
        shutil.copy(os.path.join(dataPath,data),os.path.join(imagesFolderPath,data))
    elif data.endswith('.txt') :
        shutil.copy(os.path.join(dataPath,data),os.path.join(labelsFolderPath,data))
```
