```python
from keras.utils import to_categorical
from keras import layers
from keras import models
from keras import optimizers
import matplotlib.pyplot as plt
import os , shutil
```

```python
base_dir = "./cats_and_dogs"
source_path = os.path.join("../kagglecatsanddogs_3367a/PetImages/")

cat_src_dir = os.path.join(source_path,'Cat')
dog_src_dir = os.path.join(source_path,'Dog')

if os.path.exists(base_dir) :
    shutil.rmtree(base_dir)
os.makedirs(base_dir,exist_ok=True)

train_dir = os.path.join(base_dir,'train')
test_dir = os.path.join(base_dir,'test')
validation_dir = os.path.join(base_dir,'validation')

folders = [train_dir,test_dir,validation_dir]

for item in ['cats','dogs'] :
    folders.append(os.path.join(train_dir,item))
    folders.append(os.path.join(test_dir,item))
    folders.append(os.path.join(validation_dir,item))
    
for path in folders :
    os.makedirs(path,exist_ok=True)
```
